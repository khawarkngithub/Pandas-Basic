## 2. Creating a DataFrame from a Dictionary

A dictionary can be passed to `pd.DataFrame()`. Each dictionary key becomes a column, and each list supplies the values for that column.

```python
import pandas

mydataset = {
  'cars': ["BMW", "Volvo", "Ford"],
  'passings': [3, 7, 2]
}

myvar = pandas.DataFrame(mydataset)

print(myvar)
```

## 5. Creating a Series from a List

A `Series` is a one-dimensional Pandas object. A list can be converted directly, and Pandas creates numeric labels starting at `0` by default.

```python
# Create a simple Pandas Series from a list:
import pandas as pd

a = [1, 7, 2]

myvar = pd.Series(a)

print(myvar)
```

### Selecting a Series Value by Position

With the default index, square brackets select the value associated with index label `0`.

```python
print(myvar[0])
```

## 6. Assigning Custom Series Index Labels

An `index` list can replace the default numeric labels.

```python
import pandas as pd

a = [1, 7, 2]

myvar = pd.Series(a, index=["x", "y", "z"])

print(myvar)
```

### Selecting a Series Value by Label

The custom label `"y"` selects the value associated with that index label.

```python
print(myvar["y"])
```

## 7. Creating a Series from a Dictionary

A dictionary can create a Series. By default, dictionary keys become index labels and dictionary values become the Series values.

```python
# Create a simple Pandas Series from a dictionary:
import pandas as pd

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar)
```

### Duplicate Example: Selecting Dictionary Keys

This repeats dictionary-to-Series creation but passes an index containing only `day1` and `day2`. The missing key `day3` is excluded by the requested index.

```python
# Create a Series using only data from "day1" and "day2":
import pandas as pd

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories, index=["day1", "day2"])

print(myvar)
```

## 8. Creating a DataFrame from Column Lists

This example creates a DataFrame with `calories` and `duration` columns from lists stored in a dictionary.

```python
# Create a DataFrame from two Series:
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

myvar = pd.DataFrame(data)

print(myvar)
```

### Duplicate Example: DataFrame Practice with a Named Variable

This repeats the preceding DataFrame example. It assigns the DataFrame to `df`, which makes the following row-selection examples possible.

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

#load data into a DataFrame object:
df = pd.DataFrame(data)

print(df)
```

## 9. Selecting Rows with `.loc`

`.loc` selects rows using index labels. With the default numeric index, `df.loc[0]` selects the first row.

```python
print(df.loc[0])
```

### Selecting Multiple Rows

Passing a list of labels to `.loc` selects multiple rows.

```python
print(df.loc[[0, 1]])
```

## 10. DataFrame with Custom Index Labels

The `index` argument assigns `day1`, `day2`, and `day3` as row labels instead of `0`, `1`, and `2`.

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

df = pd.DataFrame(data, index=["day1", "day2", "day3"])

print(df)
```

### Selecting a Row with a Custom Label

`df.loc["day2"]` returns the row labeled `day2`.

```python
print(df.loc["day2"])
```
